데이터분석 작업을 용이하게 해주는 파이썬 모듈(라이브러리)
> 2차원 표형태의 데이터를 아주 편하게 처리해주는 모듈    
> 데이터분석 할때 n차원 행렬의 구조로 용이하게 만들어주는 모듈 *(엑셀, csv, tsv 데이터 처리에 용이)* 

```python
#0. pandas 모듈 사용
import pandas as pd
```

### 판다스의 데이터 구조

- **Series 시리즈**   
순차데이터를 저장하는 1차원 배열로 구성
>예 : [10,20,30]

- **DataFrame 데이터프레임**   
행렬로 구성된 2차원 배열로 구성. 각 열에 서로 다른 타입의 데이터로 구성 가능
>예 : [ ['sam',20],['robin',25] ]

#### 1) Series 시리즈  
1차원 배열 구조

```python
aa= pd.Series([10,20,30])
print(aa)  #인덱스번호가 자동부여, 자료형(dtype)이 표시됨
```

```python
#시리즈의 출력을 print()없이 시도해도 결과는 같음
aa
```

```python
#실수형 데이터 시리즈
bb= pd.Series([3.14, 5.25, 7.0])
print(bb)
```

```python
#문자형 데이터 시리즈
cc= pd.Series(['sam','robin','hong']) #pandas에서는 문자형의 자료형을 object
print(cc) 
```

```python
#만약, 타입이 다른 데이터를 시리즈로 만들면?
dd= pd.Series([100,3.14,'park'])
print(dd)
```

```python
#시리즈를 만들때 자료형을 미리 고정할 수 있음.
#ee= pd.Series([100,200,'NaN'], dtype='int64') #error
ee= pd.Series([100,200,300], dtype='int64') #OK
print(ee)
```

```python
#인덱스번호를 다른 식별자로 지정할 수 있음.
ff= pd.Series(['seoul','tokyo','paris'], index=['korea','japan','france'])
print(ff)
```

```python
#특정 데이터를 추출하는 인덱싱에 번호가 아닌 식별자를 사용
ff['france']
```

```python
# 시리즈의 자료형 확인
print( type(ff) )
```

```python
# 시리즈의 요소개수(모양)을 알 수 있는 속성
ff.shape
```
