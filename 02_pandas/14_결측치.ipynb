## 결측치
- 결측치(Missing Values)란 데이터셋에서 일부 값이 누락된 것
- 결측치가 있으면 데이터분석 결과를 왜곡할 수 있기에 분석전에 미리 전처리 해 놓아야 함.

```python
import pandas as pd
df = pd.read_csv('./data_set/pandas.csv', encoding='cp949') #제공한 csv 파일이 'cp949'로 되어 있음.
df
```

```python
# 결측치 존재 확인
df.isna() # 각 cell마다 결측치 여부를 알려줌. 데이터프레임으로 결과 줌 
```

```python
df.isnull()
```

```python
# 결측치의 개수 확인
df.isna().sum()
```

```python
# 결측치의 비율이 별로 없다면 보통은 그 행을 제거한 후 분석을 함.
# 결측치의 비율 확인 - 산수
(df.isnull().mean()*100).round(2)
```

```python
# 몸무게 컬룸에서 결측치가 존재하는 행을 필터링하여 추출
df[ df['몸무게'].isna() ] #불리언 인덱싱
```

```python
# '몸무게', '급여' 컬룸의 결측치가 존재하는 행을 추출
df[  df['몸무게'].isnull() | df['급여'].isna()  ]
```

```python
# 반대로 결측치가 없는 행들만 추출
df[ df['몸무게'].notna() & df['급여'].notnull() ]
```

## 결측치 처리 방법 9가지

```python
#1. 결측치가 존재하는 컬룸을 계산해서 제외하기 위해 제거 drop
df.drop(columns= ['몸무게','급여']) #원본은 안바뀜
```

```python
# 위 방법은 몸무게와 급여 컬룸이 필요없을때..중요한 변수가 아닐때..
# 컬룸 제거 말고.. 결측치를 가진 관측값(행row)을 제거
df.drop(index=[1,3])
```

```python
# 행의 index 번호를 찾기 어려우니..결측치가 있는 행을 알아서 찾아서 제거..
df.dropna()
```

```python
#2. 0으로 채우기
df.fillna(0) #결측값을 찾아서 0으로 채워줌.
```

```python
#3. 평균값으로 채우기 -- 몸무게 컬룸만
df[['몸무게']].fillna( df['몸무게'].mean() )
```

```python
#4. 중위값 채우기 -- 급여
df[['급여']].fillna(df['급여'].median())
```

```python
#2. 0으로 채우기 - 구매금액..
df= pd.read_csv('./data_set/결측치 처리/2.purchase_amount.csv')
df
```

```python
df['구매금액']= df['구매금액'].fillna(0)
df
```

```python
# 방문횟수에 따른 구매금액 분석
df['구매금액'].sum()/df['방문횟수'].sum()
```

```python
#3. 평균으로 채우기 -- 시험성적(결측값이 반의 평균에 근접. 정규분포...가정)
df= pd.read_csv('./data_set/결측치 처리/3.exam_score.csv')
df
```

```python
# 평균으로 결측값으로 채우기 -- 표본데이터가 많을 수록 정규분포에 가까워짐.
mean_score= df['수학점수'].mean()
df['수학점수']=df['수학점수'].fillna(mean_score)
df
```

```python
# 기초 통계 분석
df['수학점수'].describe()
```

```python
#4. 결측치를 중앙값으로 채우기 ( 왜도(이상치 데이터)가 있을때.. 월소득 )
df = pd.read_csv('./data_set/결측치 처리/4.income.csv')
df
```

```python
# 데이터에 이상치(outlier)가 있는지 여부 - 탐지방식 : IQR(사분위), z-score 등

#방법1. z-score 가 3이상 
#표준점수(z-score) = (데이터-평균:내 편차)/표준편차  > 3 면 이상함.
( df['월소득']-df['월소득'].mean() ) / df['월소득'].std()
# 기준 의미
# |z-score| > 3 이상치(처리 대상)
# |z-score| > 2 약간 이상치(관심)
# |z-score| > 1 양호한 수준

# |z-score| 만으로 모든 이상치를 탐지하지 못할 수 있음..IQR도 함께 검정.
```

```python
#방법2. IQR(Inter Quartile Range) 규칙
# IQR = Q3-Q1
# 이상치의 하안선 : Q1 - 1.5*IQR
# 이상치의 상한선 : Q3 + 1.5*IQR
Q1 = df['월소득'].quantile(0.25)
Q3 = df['월소득'].quantile(0.75)
IQR= Q3-Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

lower, upper
```

```python
df[ (df['월소득']<lower) | (df['월소득']>upper) ]
#고소득 이상치 존재... -- 평균이 왜곡됨
```

```python
#이상치가 있을때.. 결측치 채우는 방법 - 중위값
df['월소득'] = df['월소득'].fillna( df['월소득'].median() )
df
```

```python
#4. 최빈값(빈독수가 많음)으로 채우기 -- 범주형 데이터일때..유용
df= pd.read_csv('./data_set/결측치 처리/5.customer_gender.csv')
df
```

```python
# 성별중에서 더 많은 성별이 누군지 판별...
mode_gender = df['성별'].mode()
mode_gender
```

```python
df['성별'] = df['성별'].fillna(mode_gender[0])
df
```

```python
# 6. 그룹별 대체 -- 실무에서 가장 많이 사용. 전체평균/중앙값 이 아니라..그룹 내 평균/중앙값 사용
df= pd.read_csv('./data_set/결측치 처리/6.group_income.csv')
df
```

```python
# 성별에 따른 소득이 다른 경향이 있기에..이를 반영...
df['소득'] = df['소득'].fillna( df.groupby('성별')['소득'].transform('mean') )
df
```

```python
#7. 예측모델 기반 대체...
df= pd.read_csv('./data_set/결측치 처리/7.house_price.csv')
df
```

```python
# 가격을 결정하는 요인이 다른 집들의 가격보다.. 내집의 특성들(면적, 방개수)에 의해 결정되는 데이터
# 가격의 요인이 된 특성값들과의 관계성을 파악!
# 면적,방개수에 따라 가격이 어떻게 형성되는지를 파악하고 이를 기반으로 결측값을 채우기
# 단순 비율로 예측하기 어려움..
# 그러니..이 비율을 머신러닝 알고리즘을 이용하여 찾도록....
# 이때 사용하는 알고리즘. [선형회귀 알고리즘]

# 용어
# 값을 결정하는 특성(요인)값을 '독립변수'라고 부름.
# 요인(특성)에 의해 결정되는 값을 '종족변수' 라고 부름 (우리가 찾고자하는 결과(예측)값)

# 머신러닝 알고리즘을 이용하여 독립변수(면적,방개수)에 따른 종족변수(가격)을 예측할 수 있음.

# 기존 (면적,방개수)에 따른 (가격) 정보를 머신러닝 알고리즘으로 훈련(학습)하면 그 패턴을 찾아냄
# 그 패턴을 기반으로 새로운 (면적, 방개수)를 주었을때 (가격)을 예측해줌.
```

```python
# 머신러닝 알고리즘에 훈련시킬 데이터 추출 (결측치가 없는 데이터)
train_df = df.dropna()

#2. 예측하고자 하는 데이터 추출 ( 종속변수가 비어있는 데이터 )
test_df= df[ df['가격'].isnull() ]  
```

```python
#3. 선형회귀 알고리즘을 구현한 모델 사용 [대표적인 머신러닝 모듈 scikit learn]
from sklearn.linear_model import LinearRegression

# 선형회귀 모델 객체 생성
model= LinearRegression()

# 모델 훈련
model.fit(train_df[['면적','방개수']], train_df['가격']) #(독립변수(요인), 종속변수(결과))

# 학습된 모델에게 예측을 원하는 데이터의 독립변수(특성값들)를 주면 종속변수값을 예측해줌
prices= model.predict( test_df[['면적','방개수']] )
print(prices)
```

```python
#결측값을 이 예측값으로 채우기
df['가격'].fillna(prices[0])
```

```python
# 모델의 예측값을 검증1. 이미 알고있는 정답데이터에 적용
# 0 H01 45   2   220000.0
# 1 H02 60   3   300000.0
model.predict( train_df[['면적','방개수']] ) #훈련용 데이터셋으로 확인하여 예측값과 실제값 비교
```

```python
#면적과 가격사이의 상관관계 정도를 알 수 있음. 상관계수 (1이면 완전 관련있는것)
df[['면적','방개수','가격']].corr()
```
